```python
from qdrant_client import QdrantClient
from qdrant_client.http.models import Distance, VectorParams
```

```python
from qdrant_client import QdrantClient
from dotenv import load_dotenv
import os

load_dotenv()
QDRANT_URL = os.getenv("QDRANT_URL")
QDRANT_API_KEY = os.getenv("QDRANT_API_KEY")

client = QdrantClient(
    url=QDRANT_URL, 
    api_key=QDRANT_API_KEY
)

# client = QdrantClient(
#     url="http://localhost:6333"
#     #path='db/qdrant_db',
#     )
```

```python
client.create_collection(
    collection_name="demo_collection",
    vectors_config=VectorParams(size=384, distance=Distance.COSINE),
)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

model_name = "sentence-transformers/all-MiniLM-L6-v2"
model_kwargs = {'device': 'mps'}
encode_kwargs = {'normalize_embeddings': True, 'batch_size':1024}

embeddings_model = HuggingFaceEmbeddings(
    model_name=model_name,
    model_kwargs=model_kwargs,
    encode_kwargs=encode_kwargs,
)
```

```python
df.info()
```

```python
from uuid import uuid4
from langchain_core.documents import Document

# Создаем документы для Qdrant с UUID
documents = []
id_mapping = {}  # Сохраним соответствие UUID -> original_id

for _, row in df.iterrows():

    content = row['description']


    # Метаданные
    metadata = {
        "tvshow_title": row.get('tvshow_title', ''),
        "year": row.get('year', ''),
        "rating": row.get('rating', ''),
        "seasons": row.get('seasons', ''),
        "series": row.get('series', ''),
        "genres": row.get('genres', ''),
        "director": row.get('director', ''),
        "actors": row.get('actors', ''), #row['actors']  # СПИСОК!
        "countries": row.get('countries', ''),
        "page_url": row.get('page_url', ''),
        "image_url": row.get('image_url', '')        
    }

    documents.append(Document(page_content=content, metadata=metadata))


uuids = [str(uuid4()) for _ in range(len(documents))]

print(f"✅ Создано {len(documents)} документов с UUID")
```

```python
documents[0]
```

```python
from langchain_qdrant import QdrantVectorStore

vector_store = QdrantVectorStore(
    client=client,
    collection_name="demo_collection",
    embedding=embeddings_model
)
```

```python
from tqdm import tqdm

batch_size = 256
total_batches = (len(documents) + batch_size - 1) // batch_size


with tqdm(total=total_batches, desc="Добавление батчей в Qdrant") as pbar:
    for i in range(0, len(documents), batch_size):
        batch_docs = documents[i:i+batch_size]
        batch_ids = uuids[i:i+batch_size]

        vector_store.add_documents(documents=batch_docs, ids=batch_ids)
        pbar.update(1)

print(f"✅ {len(documents)} документов добавлено в Qdrant!")
```

```python
from qdrant_client.models import Filter, FieldCondition, MatchAny

my_filter = Filter(
    must=[
        FieldCondition(
            key="metadata.genres",  
            match=MatchAny(any=["Horror", 'Thriller'])
        ),

        FieldCondition(
            key="metadata.year",
            match=MatchAny(any=["2025"])
        ),
    ],

    must_not=[
        FieldCondition(
            key="metadata.genres", 
            match=MatchAny(any=["Drama"]))
    ]
)

query = "Family"
results_with_scores = vector_store.similarity_search_with_score(
    query,
    filter=my_filter,
    k=7
)

for i, (doc, score) in enumerate(results_with_scores):
    print(f"\n--- Result {i+1} ---")

    doc_id = doc.metadata.get('_id', 'Not ID')

    print(f"🆔 ID в базе: {doc_id}")
    print(f"📊 Similarity Score: {score:.4f}")
    print(f"Title: {doc.metadata.get('tvshow_title', 'Not specified')}")
    print(f"Year: {doc.metadata.get('year', 'Not specified')}")
    print(f"Rating: {doc.metadata.get('rating', 'Not specified')}")
    print(f"Seasons: {doc.metadata.get('seasons', 'Not specified')}")
    print(f"Series: {doc.metadata.get('series', 'Not specified')}")
    print(f"Genres: {doc.metadata.get('genres', 'Not specified')}")
    print(f"Description: {doc.page_content[:400]}...")
```

```python
def format_docs(docs):
    """Форматирует документы для передачи в промпт"""
    formatted = []

    for i, doc in enumerate(docs, 1):
        metadata = doc.metadata

        vacancy_info = f"""
        === TV Series {i} ===
        Title: {metadata.get('tvshow_title', 'Not specified')}
        Year: {metadata.get('year', 'Not specified')}
        Rating: {metadata.get('rating', 'Not specified')}
        Seasons: {metadata.get('seasons', 'Not specified')}
        Series: {metadata.get('series', 'Not specified')}
        Genres: {metadata.get('genres', 'Not specified')}

        Description: {doc.page_content[:300]}...
        """

        formatted.append(vacancy_info)

    return "\n".join(formatted)

print("✅ Функция форматирования создана")
```

```python
from langchain_core.prompts import ChatPromptTemplate

rag_prompt = ChatPromptTemplate.from_messages([
    ("system", """Ты эксперт-аналитик рынка сериалов и ТВ-шоу с многолетним опытом и отличным чувством юмора! 🎥
    Твоя задача - проанализировать предоставленные сериалы и дать профессиональную оценку с долей иронии.
    
    Если пользователь пишет на русском языке или на каком-то другом языке, кроме английского, то пиши, что ты работаешь только с английскими текстами и помочь ему не можешь! ЕСЛИ НА АНГЛИЙСКОМ, ТО ОТВЕЧАЙ НА АНГЛИЙСКОМ!!!
    Стиль анализа:
    - Проводи глубокий анализ, но с легкой иронией над ТВ-реалиями
    - Используй сериальные мемы и шутки там, где это уместно
    - Подмечай забавные особенности сериалов (завышенные рейтинги, сюжетные дыры, смешные формулировки в описаниях)
    - Структурируй ответ с эмодзи и забавными комментариями
    - Отвечай на английском языке живым, но профессиональным тоном
    - Для каждого сериала указывай название на английском
    
    Сначала напиши свои мысли в тегах <think>...</think>, затем дай итоговый ответ. Используй только контекст базы.
     
    Помни: юмор должен быть смешным и немного оскорбительным. Цель - сделать анализ интересным!
    Если среди сериалов есть что-то особенно забавное - обязательно это отметь! 😂"""),
    ("human", """📺 ДАННЫЕ ДЛЯ АНАЛИЗА (или как говорят в сериалах - "сюжетные твисты"):
{context}
🎯 ЗАПРОС НА ЭКСПЕРТИЗУ: {question}""")
])
```

```python
retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 5}
)
```

```python
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# Создаем RAG цепочку
rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough()
    } # словарь, в котором ключи - это переменные, которые будут переданы в промпт
    | rag_prompt # промпт для RAG
    | llm # тут можно поставить любую llm-модель
    | StrOutputParser() # для вывода ответа в читаемом виде
)

print("✅ RAG цепочка создана")
```

```python
# Тестируем
question = "i like drama"
try:
    answer = rag_chain.invoke(question)
    print("🔍 ОТВЕТ RAG:")
    print(answer)
except Exception as e:
    print(f"❌ Ошибка: {e}")
```
